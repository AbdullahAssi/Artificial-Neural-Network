```python
import numpy as np
import tensorflow as tf
import tensorflow.keras as keras
from tensorflow.keras import layers
import matplotlib.pyplot as plt

# Load the data
np.random.seed(45)
x = np.random.rand(1000, 1)
y = 2.5 * x + 5 + np.random.randn(1000, 1) * 2

#visulazing the data
plt.scatter(x, y, s=1)
plt.xlabel('Input')
plt.ylabel('Output')
plt.show()

#model 
model = keras.Sequential()
model.add(layers.Dense(64, activation='relu', input_shape=(1,)))
model.add(layers.Dense(32, activation='relu'))
model.compile(optimizer='adam', loss='mean_squared_error')
model.summary()
```

```python
#train the model
model.fit(x, y, epochs=500, batch_size=32 , verbose=0)

#plot the model
plt.plot(model.history.history['loss'])
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.show()
```

```python
#predict the model
y_pred = model.predict(x)
y_pred = y_pred.reshape(-1, 1)  # Ensure y_pred has the same shape as x

# Ensure y_pred has the same number of elements as x
if len(y_pred) > len(x):
    y_pred = y_pred[:len(x)]
elif len(y_pred) < len(x):
    y_pred = np.pad(y_pred, (0, len(x) - len(y_pred)), 'constant')

plt.scatter(x, y, s=1)
plt.scatter(x, y_pred, s=1)
plt.xlabel('Input')
plt.ylabel('Output')
plt.show()
```
